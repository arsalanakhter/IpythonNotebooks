# Knapsac Path For Robots

```python
from gurobipy import *
import random as rnd
import math
import pprint
import numpy as np
import pixiedust
#import plotly.plotly as py
import plotly.offline as py
import plotly.graph_objs as go
py.init_notebook_mode()
```

We'd attempt to create a path for a robot, given the robot's fuel, in such a way that the start and end nodes of the path are fixed. We plan to maximize the number of nodes visited by the robot.

For that, lets create the sets

### Sets

- $W$: Set of Worker robots. For now, we'll just keep one robot
- $T$: Set of Tasks. Lets consider 100 tasks disperesed in an arena $[0, 100]$ x $[0, 100]$.
- $S$: Set of start locations. For now, we'll just have one.
- $E$: Set of End Locations. For now, we'll just have one.

```python
W = ["W1"]
T = [ "T"+str(i) for i in range(10)]
S = ["S"]
E = ["E"]
```

### Parameters

- $f_{w_i}$: Fuel for each worker robot.

```python
f_w_i = 300
```

Lets also define task locations.

```python
T_loc = {task: (100*rnd.random(), 100*rnd.random()) for task in T }
pprint.pprint(T_loc)
```

Lets define the start location and end location

```python
S_loc = {loc: (100*rnd.random(), 100) for loc in S }
E_loc = {loc: (100*rnd.random(), 100) for loc in E }

pprint.pprint(S_loc)
pprint.pprint(E_loc)
```

```python
N_loc = {**S_loc, **T_loc, **E_loc}
pprint.pprint(N_loc)
```

Let's define nodes to be $N$ = $S \cup T \cup E$. 
Now, let's create the arcs.

```python
N = S + T + E
arcs = [(i,j) for i in N for j in N if i!=j]
pprint.pprint(N)
```

Add weights to the arcs

```python
iter_arcs = iter(arcs)
arcs = {t: np.linalg.norm(np.array(N_loc.get(t[0]))-np.array(N_loc.get(t[1]))) for t in iter_arcs} 
pprint.pprint(arcs)
```

### Decision Variables
Lets assume we have an edge going from each node to each node. Hence we need to decide which edges to keep. So the decision variables are just the arcs that need to be visited, among all the arcs.

- $x_{ij} \in \{0,1\}$ : Binary.

Lets add these arcs to the  Gurobi model

```python
model = Model('FuelConstrainedRobots')
x = model.addVars(arcs, name="x", vtype=GRB.BINARY)

# Print variable Names
#model.optimize()
#print("\n\nVariable Names")
#print("--------------")
#for v in model.getVars():
#    print(v.VarName)
```

### Objective Function
Our objective is to maximize the number of arcs visited by the worker robot. Hence

$$max \sum_{i\in N} \sum_{j \in N}  x_{ij}  \quad  i \neq j     \tag{1}    $$

where $c_{ij}$ is the cost of going from node $i$ to node $j$

```python
objFun = quicksum(arcs.get(nodePair)*x[nodePair] for nodePair in arcs)
model.setObjective(objFun, GRB.MAXIMIZE)
```

### Constraints
Our Constraints are as follows:

The worker robot should start at the start node. Which means that an edge should go out of the start node, but not come into start node.

$$  \sum_{j\in N \setminus N_s}   x_{sj} = 1  \tag{2}$$
$$  \sum_{j\in N \setminus N_s}   x_{js} = 0  \tag{2}$$

```python
c21 = model.addConstr(quicksum(x[nodePair] for nodePair in arcs if nodePair[0]=='S') == 1, name="c21")
c22 = model.addConstr(quicksum(x[nodePair] for nodePair in arcs if nodePair[1]=='S') == 0, name="c22")
#model.write("x.lp")
# Verified by inspection
```

The worker robot should end at the end node. Which means that an edge should come into the end node, but not go out of the end node.


$$  \sum_{i\in N \setminus N_e}   x_{ie} = 1 \tag{3}$$
$$  \sum_{i\in N \setminus N_e}   x_{ei} = 0 \tag{3}$$

```python
c31 = model.addConstr(quicksum(x[nodePair] for nodePair in arcs if nodePair[1]=='E') == 1, name="c31")
c32 = model.addConstr(quicksum(x[nodePair] for nodePair in arcs if nodePair[0]=='E') == 0, name="c32")
#model.write("x.lp")
# Verified by inspection
```

Each node should only be visited either once or never (except the start and end nodes).

$$\sum_{j \in T} x_{ij} <= 1 \quad \forall i \in T \tag{4}$$


$$\sum_{i \in T} x_{ij} <= 1 \quad \forall j \in T \tag{5}$$

```python
c4 = model.addConstrs((quicksum(x[i,j] for i in T if i!=j) <= 1 for j in T), name="c4")
c5 = model.addConstrs((quicksum(x[i,j] for j in T if i!=j) <= 1 for i in T), name="c5")
#model.write("x.lp")
# Verified by inspection
```

Ensure that when a robot enters a node, it leaves the node (except start and end nodes)

$$\sum_{i \in S \cup T} x_{ih} = \sum_{j \in T \cup E} x_{hj} \quad \forall h \in T \tag{6}$$

```python
c6 = model.addConstrs(((quicksum(x[i,h] for i in S+T if i!=h)) == 
                        (quicksum(x[h,j] for j in T+E if j!=h)) for h in T), name="c6")
model.write("x.lp")
# Verified by inspection
```

Ensure that total distance covered by the robot is less than its fuel (assuming consumtion of one unit of fuel per unit of time)

$$\sum_{i \in N} \sum_{j \in N} c_{ij}x_{ij} \leq f_{w}$$

```python
expr = LinExpr(0.0)
for nodePair in arcs:
    expr.addTerms(arcs.get(nodePair), x[nodePair])
# Add the f_w_i value
expr.add(1, -f_w_i)

c7 = model.addConstr(expr <= 0, name="c7")
#model.write("x.lp")
```

Eliminate Self loops

$$\sum_{i,j \in S, i \neq j} x_{ij} \leq |S| - 1 \quad \forall S \subset T, S \neq \phi$$

```python
# Callback - use lazy constraints to eliminate sub-tours

def subtourelim(model, where):
    if where == GRB.callback.MIPSOL:
        selected = []
        print("Selected:")
        print(selected)
        print('-------')
        # make a list of edges selected in the solution
        for i in model._vars:
            sol = model.cbGetSolution(i)
            print(sol)
            if sol > 0.5:
                selected += [i]
        # find the shortest cycle in the selected edge list
        # print total number of edges sent to subtour(selected)
        print('Edges sent for subtour elimination: '+str(selected))
        # Compute the set of nodes being sent
        nodesInSelected = set()
        for edge in selected:
            # Extract the nodes in this edge
            x,y = edge.VarName.split(',')
            x = x.replace("x[", "")
            y = y.replace("]", "")            
            print(x)
            print(y)
            nodesInSelected.add(x)
            nodesInSelected.add(y)
        print('Nodes sent for subtour elim: ' + str(list(nodesInSelected)))
        tour = subtour(selected)
        print("Tour:")
        print(tour)
        # If length of tour is less than total no of arcs-1
        if len(tour) < len(nodesInSelected):
            # add a subtour elimination constraint
            expr = 0
            for i in range(len(tour)):
                for j in range(i+1, len(tour)):
                    print('tour[i]: '+ tour[i])
                    print('tour[j]: ' + tour[j])
                    for k, l in enumerate(model._vars):
                        print('k: '+str(k))
                        print('l: '+(l.VarName))
                        print(str(model._vars[k]))
                        if 'x['+ str(tour[i]) + ',' + str( tour[j])+ ']' == str(l.VarName):
                            print("Our Var")
                            print(model._vars[k])
                            req_var = model._vars[k]
                   # print([i for i, j in enumerate(model._vars) if j.VarName == str(x[tour[i], tour[j]])])
                    
                    #print(model.getVarByName((x[tour[i], tour[j]])))
                    #print(model._vars[model._vars.index([x[tour[i], tour[j]]])])
                    #print(type(i))
                    #print(type(j))
                    expr += req_var
                    print('expr:')
                    print(expr)
                    print(len(tour))
            model.cbLazy(expr <= len(tour)-2)

            
# Given a list of edges, finds the shortest subtour

def subtour(edges):
    visited = {i:False for i in N}
    cycles = []
    lengths = []
    selected = {i:[] for i in N}
    print(edges)
    for edge in edges:
        # Extract the nodes in this edge
        x,y = edge.VarName.split(',')
        x = x.replace("x[", "")
        y = y.replace("]", "") 
        print(x)
        print(y)
        selected[x].append(y)
        print(selected)
    while True:
        for k, v in visited.items():
            if v == False:
                current = k
                break
        thiscycle = [current]
        print("thiscycle:")
        print(thiscycle)
        while True:
            visited[current] = True
            neighbors = [x for x in selected[current] if not visited.get(x)]
            print('neighbors')
            print(neighbors)
            if len(neighbors) == 0:
                print('breaking due to 0 neighbors')
                break
            current = neighbors[0]
            thiscycle.append(current)
        cycles.append(thiscycle)
        print('cycles:')
        print(cycles)
        lengths.append(len(thiscycle))
        print('lengths')
        print(lengths)
        if sum(lengths) == len(T)+len(S)+len(E):
            break
    # Remove all the values in cycles with only one value
    cycles = [t for t in cycles if len(t) > 1]
    lengths = [t for t in lengths if t > 1]
    print('updated cycles : ')
    print(cycles)
    print('Returning with following cycles')
    #print(cycles[lengths.index(min(lengths))])
    #return cycles[lengths.index(min(lengths))]
    
    print(cycles[lengths.index(min(lengths))])
    return cycles[lengths.index(min(lengths))]



```

### Solving the Model

```python
#%%pixie_debugger
#model.optimize()
model._vars = model.getVars()
model.params.LazyConstraints = 1
model.optimize(subtourelim)
model.write("x.lp")
```

### Plotting the data
Lets plot the data and the results

```python
def taskNodesTrace(T_loc):
    taskTrace = go.Scatter(
        text=[],
        x=[],
        y=[],
        mode='markers',
        #hoverinfo='text',
        name = 'Task Locations',
        marker=dict(
            size=6,
            color='blue',
            line=dict(
                #color='rgba(217, 217, 217, 0.14)',
                width=0.5
                ),
            opacity=0.8
            )
        )
    
    for t in T_loc:
        x, y = T_loc.get(t)
        disp_text = 'NodeID: ' + t                           #+ '<br>f_left: ' + "{0:.2f}".format(f_left)
        taskTrace['x'] += tuple([x])
        taskTrace['y'] += tuple([y])
        taskTrace['text'] += tuple([disp_text])
    return taskTrace
```

```python
def startNodesTrace(S_loc):
    startTrace = go.Scatter(
        text=[],
        x=[],
        y=[],
        mode='markers',
        #hoverinfo='text',
        name = 'Start Locations',
        marker=dict(
            size=12,
            color='green',
            line=dict(
                #color='rgba(217, 217, 217, 0.14)',
                width=0.5
                ),
            opacity=0.8
            )
        )
    
    for s in S_loc:
        x, y = S_loc.get(s)
        disp_text = 'NodeID: ' + s                           #+ '<br>f_left: ' + "{0:.2f}".format(f_left)
        startTrace['x'] += tuple([x])
        startTrace['y'] += tuple([y])
        startTrace['text'] += tuple([disp_text])
    return startTrace
```

```python
def endNodesTrace(E_loc):
    endTrace = go.Scatter(
        text=[],
        x=[],
        y=[],
        mode='markers',
        #hoverinfo='text',
        name = 'End Locations',
        marker=dict(
            size=12,
            color='red',
            line=dict(
                #color='rgba(217, 217, 217, 0.14)',
                width=0.5
                ),
            opacity=0.8
            )
        )
    
    for e in E_loc:
        x, y = E_loc.get(e)
        disp_text = 'NodeID: ' + e                           #+ '<br>f_left: ' + "{0:.2f}".format(f_left)
        endTrace['x'] += tuple([x])
        endTrace['y'] += tuple([y])
        endTrace['text'] += tuple([disp_text])
    return endTrace
```

```python
def edgeTrace(T_loc, S_loc, E_loc, finalArcs):
    edge_trace = go.Scatter(
        x=[],
        y=[],
        text=[],
        line=dict(width=1,color='#888'),
        hoverinfo='none',
        showlegend=False,
        mode='lines+markers')

    N_loc = {**S_loc, **T_loc, **E_loc}
    for arc in finalArcs:
        x0, y0 = N_loc.get(arc[0])
        x1, y1 = N_loc.get(arc[1])
        edge_trace['x'] += tuple([x0, x1])
        edge_trace['y'] += tuple([y0, y1])
    print(edge_trace['x'])
    print(edge_trace['y'])
    print(len(edge_trace['x']))
    print(len(edge_trace['y']))
    return edge_trace
```
